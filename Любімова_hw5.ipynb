#Для датасету px.data.iris() побудувати наступні графіки.
#Графіки мають назву, що вирівнювання за центром.


###1.Побудуйте 3D-графік розподілу даних для датасету Iris з використанням властивостей "sepal_length," "sepal_width," та "petal_length." Кожна точка має бути розфарбована відповідно до класу квітки (setosa, versicolor, virginica).

```python
import plotly.express as px

iris_data = px.data.iris()

iris_3d = px.scatter_3d(iris_data,
                         x='sepal_length', y='sepal_width', z='petal_length',
                         color='species',
                         title='Просторовий розподіл характеристик ірисів',
                         color_discrete_sequence=px.colors.qualitative.Vivid)

iris_3d.update_layout(title={'text': 'Просторовий розподіл характеристик ірисів', 'x': 0.5})
iris_3d.show()
```

###2.Побудуйте гістограму для розподілу значень "petal_width" для різних видів квіток в датасеті Iris.

```python
hist_plot = px.histogram(iris_data, x="petal_width", color="species",
                         title="Частотний розподіл petal_width",
                         nbins=20, barmode='group')

hist_plot.update_layout(title_x=0.5, template='plotly_dark')
hist_plot.show()
```

###3.Побудуйте лінійний графік, що відображає зміну середнього значення "sepal_length" в залежності від значення "sepal_width" для кожного виду квітки (setosa, versicolor, virginica).

```python
stats = iris_data.groupby(['species', 'sepal_width'])['sepal_length'].mean().reset_index()

line_chart = px.line(stats, x="sepal_width", y="sepal_length", color="species",
                     markers=True, title="Залежність середнього sepal_length від ширини")

line_chart.update_layout(title_x=0.5)
line_chart.show()
```

#2.Для датасету  px.data.gapminder() побудувати наступні графіки

###1.Побудувати гістограму, яка відображає розподіл ВВП за країнами для року 2007. Обрати 10 країн з найбільшим ВВП та позначити їх на гістограмі з неперервною зміною кольору. Додати підписи над стовпцями та вирівняти їх горизонтально. Назва графіку має бути розташована за центром.

```python
import plotly.express as px

gd_set = px.data.gapminder().query("year == 2007")
top10_gdp = gd_set.sort_values("gdpPercap", ascending=False).head(10)

fig_gdp = px.bar(top10_gdp, x='country', y='gdpPercap',
                 color='gdpPercap',
                 color_continuous_scale='Viridis',
                 title='Найбільші показники ВВП за 2007 рік')

fig_gdp.update_traces(texttemplate='%{y:.2f}', textposition='outside')
fig_gdp.update_layout(title_x=0.5, uniformtext_minsize=8, uniformtext_mode='hide')
fig_gdp.show()
```

###2.Побудувати кругову діаграму, яка відображає розподіл ВВП за країнами Європи у 2018 році. Виберіть 5 країн з найбільшим ВВП і підпишіть діаграму за центром, щоб вона була зручною для сприйняття.

```python
euro_stats = px.data.gapminder().query("continent == 'Europe' and year == 2007")
top5_euro = euro_stats.nlargest(5, 'gdpPercap')

pie_val = px.pie(top5_euro, values='gdpPercap', names='country',
                 hole=0.3,
                 title='Провідні економіки Європи')

pie_val.update_layout(title_x=0.5)
pie_val.show()
```

###3.Побудувати  стовпчасту діаграму, яка відображає динаміку зміни кількості населення для Канади в країнах Океанії протягом років.

```python
df_pop = px.data.gapminder()
subset = df_pop[df_pop['country'].isin(['Canada', 'Australia', 'New Zealand'])]

pop_bar = px.bar(subset, x='year', y='pop', color='country',
                 title='Порівняння чисельності населення',
                 barmode='stack')

pop_bar.update_layout(title_x=0.5)
pop_bar.show()
```

#3.Побудувати лінійні графіки для тригонометричних функцій. Назву графіку вирівняти за центром

###1.На проміжку [0;5π] функція y=2 cos⁡x(x-π/4)

```python
import numpy as np
import matplotlib.pyplot as plt

points = np.linspace(0, 5 * np.pi, 1000)
values = 2 * np.cos(points - np.pi/4)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(points, values, 'tab:blue', linewidth=2)
ax.set_title('Графік $y = 2 \cos(x - \pi/4)$', fontsize=14, pad=15)
ax.grid(ls='--')
plt.show()
```

###2.На проміжку [0;2π] функція y=1/2  sin⁡(2x+π/3)

```python
x_axis = np.linspace(0, 2 * np.pi, 1000)
y_axis = 0.5 * np.sin(2 * x_axis + np.pi/3)

plt.figure(figsize=(12, 5))
plt.plot(x_axis, y_axis, color='#e74c3c', linestyle='-.')
plt.title('Графік $y = 0.5 \sin(2x + \pi/3)$', loc='center')
plt.fill_between(x_axis, y_axis, alpha=0.1, color='red')
plt.grid(True)
plt.show()
```

###3.На проміжку [-2π;2π] функція y=2tg(1/2 x+π/6)

```python
t = np.linspace(-2 * np.pi, 2 * np.pi, 2000)
ft = 2 * np.tan(0.5 * t + np.pi/6)

ft[np.abs(np.diff(ft, prepend=ft[0])) > 1] = np.nan

plt.style.use('seaborn-v0_8-muted')
plt.figure(figsize=(12, 6))
plt.plot(t, ft, 'g', label='2 tg(0.5x + π/6)')
plt.ylim(-15, 15)
plt.axvline(x=0, color='k', alpha=0.3)
plt.title('Візуалізація функції тангенса', fontsize=12)
plt.legend()
plt.show()
```
